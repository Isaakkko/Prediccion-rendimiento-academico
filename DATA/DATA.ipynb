# SCRIP PARA DESCARGAR LA DATA
## Dataset: https://archive.ics.uci.edu/dataset/320/student+performance

```python
import os
import urllib.request
import zipfile
import pandas as pd
```

```python
import os, urllib.request, zipfile
import pandas as pd

RAW_DIR       = "DATA/RAW"
PROCESSED_DIR = "DATA/PROCESSED"

os.makedirs(RAW_DIR, exist_ok=True)
os.makedirs(PROCESSED_DIR, exist_ok=True)

# ── Paso 1: Descargar ────────────────────────────────────────
zip_path = os.path.join(RAW_DIR, "student_performance.zip")
if not os.path.exists(zip_path):
    print("Descargando...")
    urllib.request.urlretrieve("https://archive.ics.uci.edu/static/public/320/student+performance.zip", zip_path)
    print("✓ Descargado")
else:
    print("✓ ZIP ya existe")

# ── Paso 2: Extraer ZIP principal → genera student.zip ───────
inner_zip = os.path.join(RAW_DIR, "student.zip")
if not os.path.exists(inner_zip):
    with zipfile.ZipFile(zip_path) as z:
        z.extractall(RAW_DIR)
    print("✓ ZIP principal extraído")
else:
    print("✓ ZIP principal ya extraído")

# ── Paso 3: Extraer student.zip → CSVs en DATA/RAW ───────────
mat_csv = os.path.join(RAW_DIR, "student-mat.csv")
if not os.path.exists(mat_csv):
    with zipfile.ZipFile(inner_zip) as z:
        z.extractall(RAW_DIR)
    print("✓ CSVs extraídos")
else:
    print("✓ CSVs ya extraídos")

# ── Paso 4: Procesar ─────────────────────────────────────────
df_mat = pd.read_csv(os.path.join(RAW_DIR, "student-mat.csv"), sep=';')
df_mat['subject'] = 'math'

df_por = pd.read_csv(os.path.join(RAW_DIR, "student-por.csv"), sep=';')
df_por['subject'] = 'portuguese'

df = pd.concat([df_mat, df_por], ignore_index=True)

print(f"\nMatemáticas: {df_mat.shape[0]} registros")
print(f"Portugués:   {df_por.shape[0]} registros")
print(f"Total:       {df.shape[0]} registros")
print(f"\nColumnas: {df.columns.tolist()}")
print(f"\nEstadísticas G3:\n{df['G3'].describe()}")

risk = df['G3'] < 10
print(f"\nEstudiantes en riesgo: {risk.sum()} ({risk.mean():.1%})")

# ── Paso 5: Guardar ──────────────────────────────────────────
df_mat.to_csv(os.path.join(PROCESSED_DIR, "student_math.csv"), index=False)
df_por.to_csv(os.path.join(PROCESSED_DIR, "student_portuguese.csv"), index=False)
df.to_csv(os.path.join(PROCESSED_DIR, "student_performance.csv"), index=False)

print(f"\n✓ Archivos guardados en {PROCESSED_DIR}")
print("¡COMPLETADO!")
```
